```python
from functools import reduce

from bluer_options import string
from bluer_options.logger.config import log_list

from bluer_ugv.host import signature
from bluer_ugv.README.ugvs.db import dict_of_ugvs
from bluer_ugv.README.ugvs.ethernet import find_server
from bluer_ugv.logger import logger

logger.info(f"{'.'.join(signature())},\nbuilt on {string.pretty_date()}")
```

---

```python
list_of_hostnames = reduce(
    lambda x,y: x+list(y),
    [info.get("computers",{}).values() for info in dict_of_ugvs.values()],
    []
)

log_list(logger,"found",list_of_hostnames,"hostname(s)")
```

```python
test_asset = []
for hostname in ["void"] + list_of_hostnames:
    test_asset.append([hostname]+list(find_server(hostname)))
```
